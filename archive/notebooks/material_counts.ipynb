# Counting unique `material` values in `c14_master_v08.xlsx`

A plain value count of the `material` column, not lowercased/split/matched against anything -
just how often each distinct raw value shows up.

```python
from pathlib import Path

import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

OUTPUT_DIR = Path('../output/material')
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
df = pd.read_excel('../data/c14_master_v08.xlsx')
print(f"{df['material'].notna().sum()} rows with material, {df['material'].isna().sum()} without, "
      f"{df['material'].nunique()} distinct values")
df['material'].head(10)
```

```python
material_counts = (
    df['material']
    .value_counts(dropna=False)
    .rename_axis('material')
    .reset_index(name='count')
)
print(f"total count: {material_counts['count'].sum()} (equals {len(df)}, the row count of the source data)")
material_counts
```

```python
material_counts.to_csv(OUTPUT_DIR / 'material_counts.csv', index=False)
print(f'Saved {len(material_counts)} rows to material_counts.csv')
```
