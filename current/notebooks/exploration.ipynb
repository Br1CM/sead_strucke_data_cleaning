### Extract `lab_id` prefixes and match against `tbl_dating_labs`

Same approach as the `lab_no` matching done in `archive/notebooks/explore.ipynb` for
`c14_master_v08.xlsx`, adapted to this dataset's `lab_id` column. The *prefix* of `lab_id`
(the letters before the dash/number, e.g. `Ua` in `Ua-49252`) should correspond to a lab's
`international_lab_id` in `tbl_dating_labs`. We keep the full `lab_id` string as-is, but use
the prefix to resolve each row to a SEAD `dating_lab_id`. The end of this section exports a
full prefix-to-lab mapping for manual review, since not every value can be resolved
automatically.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path('../..').resolve()))
from shared.resolution import common

# .env lives at the true repo root, two levels up from current/notebooks/.
engine = common.get_db_engine('../../.env')

dating_labs = pd.read_sql(
    "select dating_lab_id, international_lab_id, lab_name from public.tbl_dating_labs",
    engine,
)
print(f"{len(dating_labs)} labs in sead_staging")
dating_labs.head()
```

#### Extracting the prefix

Not every `lab_id` follows the `<prefix>-<number>` pattern, so the parser below sorts every
row into one of three categories instead of silently dropping the awkward ones:

- **`standard`** -- a normal `<letters><separator><number>` value, e.g. `Ua-49252`. A handful
  of these spell out `Okänd` (Swedish for "unknown") in place of the accession number, e.g.
  `Ua-Okänd62` or `ICA-Okänd 2` -- the lab is still known, only the specific sample number is
  missing, so `Okänd<number>` is stripped out before reading the prefix.
- **`unknown_no_lab`** -- `lab_id` is literally just `Okänd ###` with no lab code at all -- a
  genuinely unknown lab.
- **`non_standard_id`** -- ids like `17C/0902` or `14C-5261` that don't start with a letter at
  all. These look like an internal accession scheme (year + a 1-2 letter sample-type code +
  sequence number), not an international lab code, so no lab can be derived from them.

```python
import re

def lab_id_prefix(lab_id):
    s = str(lab_id)
    # "Okänd<number>" means the *sample number* is unknown, not the lab -- drop it before
    # reading the prefix, e.g. "Ua-Okänd62" -> "Ua"
    core = re.sub(r"[-\s]?[Oo]k[äa]nd\s*\d*", "", s)
    leading = re.match(r"^[A-Za-z]+", core)
    if leading:
        return leading.group(0), "standard"
    if core.strip() == "":
        return "(Okänd -- no lab given at all)", "unknown_no_lab"
    # no leading letter, but there might be one buried in there (e.g. the "C" in "17C/0902")
    embedded = re.search(r"[A-Za-z]+", core)
    tag = embedded.group(0) if embedded else "?"
    return f"~{tag} (embedded, not a leading prefix)", "non_standard_id"

parsed = df["lab_id"].apply(lab_id_prefix)
df["lab_id_prefix"] = parsed.apply(lambda t: t[0])
df["lab_id_prefix_category"] = parsed.apply(lambda t: t[1])
df["lab_id_prefix_category"].value_counts()
```

#### Matching prefixes against `international_lab_id`

Only `standard`-category prefixes are matchable at all. `international_lab_id` isn't
consistently cased against this dataset, and some codes carry a trailing `*` (the
*Radiocarbon* journal convention for a lab that's no longer active) that this dataset never
has. Match in three tiers -- exact, case-insensitive, and case-insensitive with the trailing
`*` stripped -- and record which tier each prefix matched on, so low-confidence matches stay
visible instead of being silently accepted.

```python
exact = dict(zip(dating_labs["international_lab_id"], dating_labs["dating_lab_id"]))
by_lower = {code.lower(): (code, lab_id) for code, lab_id in exact.items()}
by_lower_stripped = {code.rstrip("*").lower(): (code, lab_id) for code, lab_id in exact.items()}

def match_lab(prefix, category):
    if category != "standard":
        return pd.Series([None, None, None])
    if prefix in exact:
        return pd.Series([exact[prefix], prefix, "exact"])
    hit = by_lower.get(prefix.lower())
    if hit:
        return pd.Series([hit[1], hit[0], "case_insensitive"])
    hit = by_lower_stripped.get(prefix.rstrip("*").lower())
    if hit:
        return pd.Series([hit[1], hit[0], "stripped_star"])
    return pd.Series([None, None, None])

prefix_matches = (
    df.groupby(["lab_id_prefix", "lab_id_prefix_category"])
    .agg(n_rows=("lab_id", "size"),
         example_lab_id=("lab_id", lambda s: "; ".join(sorted(s.astype(str).unique())[:3])))
    .reset_index()
)
prefix_matches[["dating_lab_id", "matched_international_lab_id", "match_type"]] = prefix_matches.apply(
    lambda r: match_lab(r["lab_id_prefix"], r["lab_id_prefix_category"]), axis=1,
)
lab_name_by_id = dict(zip(dating_labs["dating_lab_id"], dating_labs["lab_name"]))
prefix_matches["matched_lab_name"] = prefix_matches["dating_lab_id"].map(lab_name_by_id)
prefix_matches = prefix_matches.sort_values("n_rows", ascending=False).reset_index(drop=True)
prefix_matches
```

```python
matched_mask = prefix_matches["dating_lab_id"].notna()
by_type = prefix_matches.groupby("match_type", dropna=False)["n_rows"].sum().sort_values(ascending=False)
total_matched = prefix_matches.loc[matched_mask, "n_rows"].sum()
print(f"{total_matched} / {len(df)} rows ({total_matched / len(df):.1%}) resolve to a SEAD dating_lab_id")
print()
print("rows by match_type (NaN = unresolved):")
print(by_type)
print()
print("prefixes with a real-looking lab code but no match in tbl_dating_labs (candidates to add to SEAD):")
prefix_matches[prefix_matches["lab_id_prefix"].notna() & ~matched_mask].sort_values("n_rows", ascending=False)
```

#### Possible-match suggestions and a full export for manual review

For every unresolved `standard` prefix, suggest the closest `international_lab_id` values
(via string similarity) as hints -- not auto-applied, just a head start for whoever reviews
this. Then export **one row per unique prefix/category** -- including the `non_standard_id`
and `unknown_no_lab` buckets -- as the full mapping from every `lab_id` value in the dataset
to a `dating_lab_id`, with a blank `manual_dating_lab_id` column to fill in after review, same
pattern as `landskap_counts.csv`'s export above.

```python
import difflib

all_codes = dating_labs["international_lab_id"].tolist()
unknown_lab_id = dating_labs.loc[dating_labs["international_lab_id"] == "Unknown", "dating_lab_id"].iloc[0]

def suggest_matches(row):
    if row["lab_id_prefix_category"] != "standard" or pd.notna(row["dating_lab_id"]):
        return ""
    close = difflib.get_close_matches(row["lab_id_prefix"], all_codes, n=3, cutoff=0.6)
    return "; ".join(f"{c} (dating_lab_id={exact[c]}, {lab_name_by_id[exact[c]]})" for c in close)

review = prefix_matches.copy()
review["possible_match_suggestions"] = review.apply(suggest_matches, axis=1)

review.loc[review["lab_id_prefix_category"] == "unknown_no_lab",
           ["dating_lab_id", "matched_international_lab_id", "matched_lab_name", "match_type"]] = [
    unknown_lab_id, "Unknown", lab_name_by_id[unknown_lab_id], "unknown_placeholder",
]

review["needs_review"] = (
    review["match_type"].isin(["stripped_star", "unknown_placeholder"]) | review["match_type"].isna()
)
review["manual_dating_lab_id"] = pd.NA

notes = pd.Series("", index=review.index)
notes[review["lab_id_prefix_category"] == "non_standard_id"] = (
    "lab_id doesn't follow <prefix>-<number>; the embedded letter is a sample-type code, not "
    "an international lab code -- needs manual research to identify the lab"
)
notes[review["lab_id_prefix_category"] == "unknown_no_lab"] = (
    "lab_id literally says 'Okänd' with no lab code -- pre-filled with SEAD's generic 'Unknown' "
    "lab; override manual_dating_lab_id if the real lab can be identified"
)
notes[review["match_type"] == "stripped_star"] = (
    "only matches after stripping the DB's trailing '*' and ignoring case -- confirm before trusting it"
)
review["notes"] = notes

review = review[[
    "lab_id_prefix", "lab_id_prefix_category", "n_rows", "example_lab_id", "match_type",
    "dating_lab_id", "matched_international_lab_id", "matched_lab_name",
    "possible_match_suggestions", "needs_review", "manual_dating_lab_id", "notes",
]]

OUTPUT_DIR = Path('../output/lab_id')
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
review.to_csv(OUTPUT_DIR / 'lab_id_prefix_matches.csv', index=False)
review.to_excel(OUTPUT_DIR / 'lab_id_prefix_matches.xlsx', index=False)
print(f"wrote {OUTPUT_DIR}/lab_id_prefix_matches.csv and .xlsx: {len(review)} prefixes covering "
      f"{review['n_rows'].sum()} rows, {review['needs_review'].sum()} flagged for manual review")
review
```

#### Assigning `dating_lab_id`

Join the per-prefix matches back onto every row. `unknown_no_lab` rows get SEAD's generic
`Unknown` lab by default; everything else stays `NaN` until resolved via the exported CSV
above. `dating_lab_match_type` travels along so low-confidence and unresolved rows stay easy
to filter out downstream if a stricter cut is wanted before ingestion.

```python
df = df.merge(
    prefix_matches[["lab_id_prefix", "lab_id_prefix_category", "dating_lab_id", "match_type"]].rename(
        columns={"match_type": "dating_lab_match_type"}
    ),
    on=["lab_id_prefix", "lab_id_prefix_category"], how="left",
)
df.loc[df["lab_id_prefix_category"] == "unknown_no_lab", ["dating_lab_id", "dating_lab_match_type"]] = [
    unknown_lab_id, "unknown_placeholder",
]
print(f'{df["dating_lab_id"].notna().sum()} / {len(df)} rows have a dating_lab_id')
df[["lab_id", "lab_id_prefix", "lab_id_prefix_category", "dating_lab_id", "dating_lab_match_type"]].sample(10, random_state=1)
```
