# Author full

In SEAD, tbl_biblio has the column named author whose values seem to be the author followed by the year in parenthesis

## Approach

`author` has 2020 distinct raw values with wildly inconsistent formatting: most are already `Surname, Firstname`, some are `Firstname Surname` with no comma at all, several authors get joined with `&` (or plain commas, or the Swedish word `och`), and on top of that there are typos, OCR-looking artifacts, `m fl` ("et al."), `(red)` ("ed."), and nobility particles (`von`, `von der`, ...) scattered around.

Since there are only ~2020 unique strings behind the 30k+ rows, the plan is:
1. Parse every unique author string once with a heuristic normalizer that targets the `Surname, N., Surname2, N2. & Surname3, N3.` pattern.
2. Flag anything the heuristic can't confidently reorder (rather than silently guessing) so it can be reviewed/corrected by hand.
3. Merge the normalized names back onto the full dataframe and append `(year)` to build `author_full`.

```python
unique_authors = sorted(df['author'].dropna().unique())
len(unique_authors)
```

## Step 1 — Heuristic parser

Rules, in order:
- Strip trailing `m fl` / `m. fl.` (→ " et al." suffix) and `(red)` (→ " (ed.)" suffix).
- Replace the Swedish conjunction `och` with `&`.
- If the string has commas: a single `&` always precedes the last author in this dataset, so replacing `&` with `,` and splitting on commas turns the whole thing into a flat token list. Since every author is written as exactly two tokens (`Surname, Firstname`), pairing tokens two-by-two recovers each author. An odd leftover token is checked once more in case it hides an unseparated `Surname Initial` pair (e.g. `"...,  Wallebom U"`).
- If the string has commas trailing a nobility particle (`von`, `von der`, `af`, ...) stuck to the wrong side of the pair, move the particle back onto the surname (`"Luft, Magnus von der"` → `"von der Luft, Magnus"`).
- If the string has **no comma at all**, name order is genuinely ambiguous (`"Firstname Surname"` vs `"Surname Firstname"`) — only two narrow, safe patterns are auto-fixed: a stray `.`/`;`/`:` typo'd in place of a comma (`"Edring. Anders"`), and a bare `Surname Initial` with no punctuation (`"Ebbesen K"`). Anything else with no comma is left untouched and flagged.
- First names are reduced to a single initial (`"Mattias"` → `"M."`, `"Anna-Karin"` → `"A."`).
- Anything with digits embedded in the name, or an odd token count that can't be repaired, is flagged `needs_review` and falls back to the original (suffix-stripped) text instead of a guess.

```python
import re

ET_AL_RE = re.compile(r'[,\s]*\(?\bm\.?\s*f\.?l\.?\)?\.?\s*$', re.I)
RED_RE = re.compile(r'[,\s]*\(?\s*red\.?\s*\)?\.?\s*$', re.I)
PARTICLE_RE = re.compile(r'\b(von|van|af|de)\b', re.I)
DIGIT_RE = re.compile(r'\d')
OCH_RE = re.compile(r'\s+och\s+', re.I)
NAME_WORD = r"[A-ZÅÄÖÜ][\wÀ-ÿ\-']*"
# "Surname. Firstname" / "Surname; Firstname" / "Surname: Firstname" (typo'd comma)
PSEUDO_COMMA_RE = re.compile(rf'^({NAME_WORD})[.;:]\s+({NAME_WORD})$')
# "Surname Initial" with no separator at all, e.g. "Ebbesen K" / "Adamsen Chr."
SHORT_INITIAL_RE = re.compile(rf'^({NAME_WORD})\s+([A-ZÅÄÖÜ][a-zåäöA-ZÅÄÖ]{{0,3}}\.?)$')
PARTICLE_WORDS = {'von', 'van', 'af', 'de', 'der', 'den', 'zu', 'du', 'la', 'le'}


def strip_suffixes(s):
    """Pull off trailing 'm fl' (et al.) / '(red)' (ed.) annotations, in either order."""
    et_al = False
    editor = False
    changed = True
    while changed:
        changed = False
        new_s = ET_AL_RE.sub('', s)
        if new_s != s:
            et_al, s, changed = True, new_s, True
        new_s = RED_RE.sub('', s)
        if new_s != s:
            editor, s, changed = True, new_s, True
    return s.strip(' ,.'), et_al, editor


def fix_particle_order(surname, firstname):
    """Move a trailing nobility particle (e.g. 'Magnus von der') from the firstname
    back onto the surname ('von der Luft'), which is where OCR/data-entry usually lost it."""
    words = firstname.split()
    particles = []
    while words and words[-1].lower() in PARTICLE_WORDS:
        particles.insert(0, words.pop().lower())
    if particles:
        surname = ' '.join(particles) + ' ' + surname
        firstname = ' '.join(words)
    return surname, firstname


def initialize(name_part):
    """Turn a first-name segment into an initial, e.g. 'Mattias' -> 'M.', 'Anna-Karin' -> 'A.'"""
    name_part = name_part.strip(' .')
    if not name_part:
        return ''
    first_word = re.split(r'[\s\-]', name_part)[0]
    return first_word[0].upper() + '.' if first_word else ''


def parse_no_comma_segment(segment):
    """Try to safely split a single no-comma author segment into (surname, firstname).
    Returns None if the order can't be determined with confidence."""
    segment = segment.strip()
    m = PSEUDO_COMMA_RE.match(segment) or SHORT_INITIAL_RE.match(segment)
    return (m.group(1), m.group(2)) if m else None


def build_result(original, formatted, et_al, editor, flags):
    suffix = ''
    if et_al:
        suffix += ' et al.'
    if editor:
        suffix += ' (ed.)'
    formatted_full = (formatted + suffix).strip()
    needs_review = 'needs_review' in flags
    reason = ','.join(sorted(set(f for f in flags if f != 'needs_review')))
    return original, formatted_full, needs_review, reason


def parse_author_string(raw):
    original = raw
    flags = []

    s = OCH_RE.sub(' & ', raw.strip())
    s, et_al, editor = strip_suffixes(s)
    if et_al:
        flags.append('et_al')
    if editor:
        flags.append('editor')
    if DIGIT_RE.search(s):
        flags.append('digits')
    if PARTICLE_RE.search(s):
        flags.append('particle')

    if ',' not in s:
        flags.append('no_comma')
        segments = [seg.strip() for seg in re.split(r'\s*&\s*', s) if seg.strip()]
        parsed_pairs = [parse_no_comma_segment(seg) for seg in segments]
        if all(parsed_pairs):
            authors = [f'{surname}, {initialize(firstname)}'.strip() for surname, firstname in parsed_pairs]
            formatted = authors[0] if len(authors) == 1 else ', '.join(authors[:-1]) + ' & ' + authors[-1]
        else:
            formatted = s  # can't safely determine name order
            flags.append('needs_review')
        return build_result(original, formatted, et_al, editor, flags)

    # a single '&' always precedes the final author in this dataset (verified: no multi-'&' cases)
    s_norm = re.sub(r'\s*&\s*', ', ', s)
    tokens = [t.strip() for t in s_norm.split(',') if t.strip()]

    if len(tokens) % 2 != 0:
        # repair pass: the last token may itself hide an unseparated "Surname Initial" pair
        m = SHORT_INITIAL_RE.match(tokens[-1]) if tokens else None
        if m:
            tokens = tokens[:-1] + [m.group(1), m.group(2)]

    if len(tokens) % 2 != 0:
        flags += ['odd_tokens', 'needs_review']
        return build_result(original, s, et_al, editor, flags)

    authors = []
    for i in range(0, len(tokens), 2):
        surname, firstname = tokens[i], tokens[i + 1]
        if 'particle' in flags:
            surname, firstname = fix_particle_order(surname, firstname)
        initial = initialize(firstname)
        if not initial:
            flags.append('needs_review')
        authors.append(f'{surname}, {initial}'.strip())

    formatted = authors[0] if len(authors) == 1 else ', '.join(authors[:-1]) + ' & ' + authors[-1]
    if 'digits' in flags:
        flags.append('needs_review')

    return build_result(original, formatted, et_al, editor, flags)
```

## Step 2 — Run the parser over every unique value and flag uncertain ones

The result is saved to `output/author_normalization_review.csv` — a lookup table with one row per unique raw `author` value, its parsed form, and whether it needs a human look. Rows flagged `needs_review` fall back to the original (suffix-stripped) text rather than a guessed reordering, so they're easy to spot and hand-correct in the CSV later.

```python
rows = []
for a in unique_authors:
    original, formatted, needs_review, reason = parse_author_string(a)
    rows.append({'author_raw': original, 'author_parsed': formatted, 'needs_review': needs_review, 'flag_reason': reason})

author_review_df = pd.DataFrame(rows)
author_review_df.to_csv('../output/author_normalization_review.csv', index=False, encoding='utf-8-sig')

total = len(author_review_df)
flagged = author_review_df['needs_review'].sum()
print(f'total unique authors: {total}')
print(f'flagged for review: {flagged} ({flagged/total:.1%})')
print()
print(author_review_df.loc[author_review_df['needs_review'], 'flag_reason'].value_counts())
```

A peek at what got flagged and why — mostly genuinely ambiguous name order (`no_comma`), malformed comma counts (`odd_tokens`), or garbage/OCR artifacts (`digits`):

```python
author_review_df[author_review_df['needs_review']].sample(15, random_state=1)[['author_raw', 'author_parsed', 'flag_reason']]
```

## Step 3 — Merge the normalized names back onto `df` and add the year

`author_review_df` is a one-row-per-unique-value lookup, so merging it onto `df` on the `author` column is a clean left join with no row duplication. `author_full` then appends `(publication_year)` — left off entirely when the year is missing rather than producing an empty `()`.

```python
df_new = df.merge(
    author_review_df[['author_raw', 'author_parsed', 'needs_review']],
    left_on='author', right_on='author_raw', how='left'
).drop(columns='author_raw').rename(columns={'author_parsed': 'author_normalized', 'needs_review': 'author_needs_review'})

df_new['author_full'] = df_new.apply(
    lambda r: f"{r['author_normalized']} ({int(r['publication_year'])})" if pd.notna(r['publication_year']) else r['author_normalized'],
    axis=1
)

print(f"rows flagged for manual author review: {df_new['author_needs_review'].sum()} / {len(df_new)}")
df_new[['author', 'author_normalized', 'author_needs_review', 'publication_year', 'author_full']]
```

## Result

`df_new['author_full']` now holds `Surname, N., Surname2, N2. & Surname3, N3. (year)`, built from ~2020 unique author strings normalized once and mapped back onto all 30,301 rows.

~4% of rows (`author_needs_review == True`) carry a raw author string the heuristic wasn't confident about (ambiguous `Firstname Surname` order, malformed punctuation, or OCR-looking garbage) — for those, `author_normalized` falls back to the original text. To fix them: edit `author_parsed` in `output/author_normalization_review.csv` for the flagged rows, then re-run the Step 3 merge cell to refresh `df_new`.

```python
import re

material_pattern = re.compile(r'^(?P<used>[^,(]+)\s*(?:,\s*(?P<mod_comma>.+)|\((?P<mod_paren>[^)]+)\))?$')

def split_material(value):
    if pd.isna(value):
        return pd.Series({'material_used': pd.NA, 'material_modification': pd.NA})
    match = material_pattern.match(value.strip())
    if not match:
        return pd.Series({'material_used': value.strip(), 'material_modification': pd.NA})
    modification = match.group('mod_comma') or match.group('mod_paren')
    return pd.Series({
        'material_used': match.group('used').strip(),
        'material_modification': modification.strip() if modification else pd.NA,
    })

df[['material_used', 'material_modification']] = df['material'].apply(split_material)
```

## Matching `lab_no` to SEAD's `tbl_dating_labs`

The idea: the *prefix* of `lab_no` (the letters before the dash/number, e.g. `Ua` in
`Ua-49252`) should correspond to a lab's `international_lab_id` in `tbl_dating_labs`. We keep
the full `lab_no` string as-is, but use the prefix to resolve each row to a SEAD
`dating_lab_id`. The end of this section exports a full prefix-to-lab mapping for manual
review, since not every value can be resolved automatically.

```python
dating_labs = pd.read_sql(
    "select dating_lab_id, international_lab_id, lab_name from public.tbl_dating_labs",
    engine,
)
print(f"{len(dating_labs)} labs in sead_staging")
dating_labs.head()
```

### Extracting the prefix

Not every `lab_no` follows the `<prefix>-<number>` pattern, so the parser below sorts every
row into one of three categories instead of silently dropping the awkward ones:

- **`standard`** -- a normal `<letters><separator><number>` value, e.g. `Ua-49252`. A handful
  of these spell out `Okänd` (Swedish for "unknown") in place of the accession number, e.g.
  `Ua-Okänd62` or `STOkänd 22` -- the lab is still known, only the specific number is missing,
  so `Okänd<number>` is stripped out before reading the prefix.
- **`unknown_no_lab`** -- `lab_no` is literally just `Okänd ###` with no lab code at all --
  a genuinely unknown lab.
- **`non_standard_id`** -- ids like `17C/0235` or `14C-5261` that don't start with a letter at
  all. These look like an internal accession scheme (year + a 1-2 letter sample-type code +
  sequence number), not an international lab code, so no lab can be derived from them.

```python
import re

def lab_no_prefix(lab_no):
    s = str(lab_no)
    # "Okänd<number>" means the *sample number* is unknown, not the lab -- drop it before
    # reading the prefix, e.g. "Ua-Okänd62" -> "Ua"
    core = re.sub(r"[-\s]?[Oo]k[äa]nd\s*\d*", "", s)
    leading = re.match(r"^[A-Za-z]+", core)
    if leading:
        return leading.group(0), "standard"
    if core.strip() == "":
        return "(Okänd -- no lab given at all)", "unknown_no_lab"
    # no leading letter, but there might be one buried in there (e.g. the "C" in "17C/0235")
    embedded = re.search(r"[A-Za-z]+", core)
    tag = embedded.group(0) if embedded else "?"
    return f"~{tag} (embedded, not a leading prefix)", "non_standard_id"

parsed = df["lab_no"].apply(lab_no_prefix)
df["lab_no_prefix"] = parsed.apply(lambda t: t[0])
df["lab_no_prefix_category"] = parsed.apply(lambda t: t[1])
df["lab_no_prefix_category"].value_counts()
```

### Matching prefixes against `international_lab_id`

Only `standard`-category prefixes are matchable at all. `international_lab_id` isn't
consistently cased against the excel data, and some codes carry a trailing `*` (the
*Radiocarbon* journal convention for a lab that's no longer active) that the excel side never
has. Match in three tiers -- exact, case-insensitive, and case-insensitive with the trailing
`*` stripped -- and record which tier each prefix matched on, so low-confidence matches stay
visible instead of being silently accepted.

```python
exact = dict(zip(dating_labs["international_lab_id"], dating_labs["dating_lab_id"]))
by_lower = {code.lower(): (code, lab_id) for code, lab_id in exact.items()}
by_lower_stripped = {code.rstrip("*").lower(): (code, lab_id) for code, lab_id in exact.items()}

def match_lab(prefix, category):
    if category != "standard":
        return pd.Series([None, None, None])
    if prefix in exact:
        return pd.Series([exact[prefix], prefix, "exact"])
    hit = by_lower.get(prefix.lower())
    if hit:
        return pd.Series([hit[1], hit[0], "case_insensitive"])
    hit = by_lower_stripped.get(prefix.rstrip("*").lower())
    if hit:
        return pd.Series([hit[1], hit[0], "stripped_star"])
    return pd.Series([None, None, None])

prefix_matches = (
    df.groupby(["lab_no_prefix", "lab_no_prefix_category"])
    .agg(n_rows=("lab_no", "size"),
         example_lab_no=("lab_no", lambda s: "; ".join(sorted(s.astype(str).unique())[:3])))
    .reset_index()
)
prefix_matches[["dating_lab_id", "matched_international_lab_id", "match_type"]] = prefix_matches.apply(
    lambda r: match_lab(r["lab_no_prefix"], r["lab_no_prefix_category"]), axis=1,
)
lab_name_by_id = dict(zip(dating_labs["dating_lab_id"], dating_labs["lab_name"]))
prefix_matches["matched_lab_name"] = prefix_matches["dating_lab_id"].map(lab_name_by_id)
prefix_matches = prefix_matches.sort_values("n_rows", ascending=False).reset_index(drop=True)
prefix_matches
```

### Coverage

Not quite -- we don't already have every lab. A few things stand out:

- **`ST` (3622 rows, ~12% of the dataset)** only matches on the loosest tier, to `St*`
  ("Stockholm", flagged defunct) -- plausible (Stockholm's old lab is commonly cited as
  `St-nnnn` in the literature) but the excel format (`ST3045`, no dash, all caps) doesn't
  resemble any other prefix in the sheet, so this needs a human to confirm before it's trusted
  for ingestion.
- **Genuinely missing labs** (a real-looking code with zero match in `tbl_dating_labs`, even
  loosely): `LTL`, `LuA`, `UGAMS`, `MAMS`, `CAIS`, `MKL`, `ICA`, `KlA`, `Tua` -- these would
  need new rows added to `tbl_dating_labs` to be resolvable at all.
- **`non_standard_id` rows (~576 rows)** -- grouped by their embedded letter (`C`, `P`, `W`,
  `B`, `O`, `M`, `T`, `CB`); none of these are real lab codes, so they can't resolve.
- **`unknown_no_lab` (85 rows)** -- no lab encoded at all; defaulted to SEAD's dedicated
  `Unknown` dating lab below.

```python
matched_mask = prefix_matches["dating_lab_id"].notna()
by_type = prefix_matches.groupby("match_type", dropna=False)["n_rows"].sum().sort_values(ascending=False)
total_matched = prefix_matches.loc[matched_mask, "n_rows"].sum()
print(f"{total_matched} / {len(df)} rows ({total_matched / len(df):.1%}) resolve to a SEAD dating_lab_id")
print()
print("rows by match_type (NaN = unresolved):")
print(by_type)
print()
print("prefixes with a real-looking lab code but no match in tbl_dating_labs (candidates to add to SEAD):")
prefix_matches[prefix_matches["lab_no_prefix"].notna() & ~matched_mask].sort_values("n_rows", ascending=False)
```

### Possible-match suggestions and a full export for manual review

For every unresolved `standard` prefix, suggest the closest `international_lab_id` values
(via string similarity) as hints -- not auto-applied, just a head start for whoever reviews
this. Then export **one row per unique prefix/category** -- including the `non_standard_id`
and `unknown_no_lab` buckets -- as the full mapping from every `lab_no` value in the dataset
to a `dating_lab_id`, with a blank `manual_dating_lab_id` column to fill in after review.

```python
import difflib

all_codes = dating_labs["international_lab_id"].tolist()
unknown_lab_id = dating_labs.loc[dating_labs["international_lab_id"] == "Unknown", "dating_lab_id"].iloc[0]

def suggest_matches(row):
    if row["lab_no_prefix_category"] != "standard" or pd.notna(row["dating_lab_id"]):
        return ""
    close = difflib.get_close_matches(row["lab_no_prefix"], all_codes, n=3, cutoff=0.6)
    return "; ".join(f"{c} (dating_lab_id={exact[c]}, {lab_name_by_id[exact[c]]})" for c in close)

review = prefix_matches.copy()
review["possible_match_suggestions"] = review.apply(suggest_matches, axis=1)

review.loc[review["lab_no_prefix_category"] == "unknown_no_lab",
           ["dating_lab_id", "matched_international_lab_id", "matched_lab_name", "match_type"]] = [
    unknown_lab_id, "Unknown", lab_name_by_id[unknown_lab_id], "unknown_placeholder",
]

review["needs_review"] = (
    review["match_type"].isin(["stripped_star", "unknown_placeholder"]) | review["match_type"].isna()
)
review["manual_dating_lab_id"] = pd.NA

notes = pd.Series("", index=review.index)
notes[review["lab_no_prefix_category"] == "non_standard_id"] = (
    "lab_no doesn't follow <prefix>-<number>; the embedded letter is a sample-type code, not "
    "an international lab code -- needs manual research to identify the lab"
)
notes[review["lab_no_prefix_category"] == "unknown_no_lab"] = (
    "lab_no literally says 'Okänd' with no lab code -- pre-filled with SEAD's generic 'Unknown' "
    "lab; override manual_dating_lab_id if the real lab can be identified"
)
notes[review["match_type"] == "stripped_star"] = (
    "only matches after stripping the DB's trailing '*' and ignoring case -- confirm before trusting it"
)
review["notes"] = notes

review = review[[
    "lab_no_prefix", "lab_no_prefix_category", "n_rows", "example_lab_no", "match_type",
    "dating_lab_id", "matched_international_lab_id", "matched_lab_name",
    "possible_match_suggestions", "needs_review", "manual_dating_lab_id", "notes",
]]

out_dir = "../output/lab_no"
os.makedirs(out_dir, exist_ok=True)
review.to_csv(f"{out_dir}/lab_no_prefix_matches.csv", index=False)
review.to_excel(f"{out_dir}/lab_no_prefix_matches.xlsx", index=False)
print(f"wrote {out_dir}/lab_no_prefix_matches.{{csv,xlsx}}: {len(review)} prefixes covering "
      f"{review['n_rows'].sum()} rows, {review['needs_review'].sum()} flagged for manual review")
review
```

### Assigning `dating_lab_id`

Join the per-prefix matches back onto every row. `unknown_no_lab` rows get SEAD's generic
`Unknown` lab by default; everything else stays `NaN` until resolved via the exported CSV
above. `dating_lab_match_type` travels along so low-confidence and unresolved rows stay easy
to filter out downstream if a stricter cut is wanted before ingestion.

```python
df = df.merge(
    prefix_matches[["lab_no_prefix", "lab_no_prefix_category", "dating_lab_id", "match_type"]].rename(
        columns={"match_type": "dating_lab_match_type"}
    ),
    on=["lab_no_prefix", "lab_no_prefix_category"], how="left",
)
df.loc[df["lab_no_prefix_category"] == "unknown_no_lab", ["dating_lab_id", "dating_lab_match_type"]] = [
    unknown_lab_id, "unknown_placeholder",
]
print(f'{df["dating_lab_id"].notna().sum()} / {len(df)} rows have a dating_lab_id')
df[["lab_no", "lab_no_prefix", "lab_no_prefix_category", "dating_lab_id", "dating_lab_match_type"]].sample(10, random_state=1)
```
